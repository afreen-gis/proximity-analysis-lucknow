```python
import pandas as pd
```

```python
import os
folder_name = 'data'
filename = 'worldcities.csv'
full_path = os.path.join(folder_name,filename)
```

```python
df = pd.read_csv(full_path)
```

FILTERING

```python
needed_country = 'India'
filtered = df[df['country']== needed_country]
filtered
```

```python
city_needed = 'Lucknow'
filtered = df[df['city_ascii']== city_needed]
filtered
```

```python
country_df = df[df['country']== needed_country].copy()
```

```python
city_needed_coordinates = (filtered.iloc[0]['lat'],filtered.iloc[0]['lng'])
print(city_needed_coordinates)
```

CALCULATING DISTANCE


```python
from geopy import distance
def calculate_distance_from(row):
    city_coordinates = (row['lat'], row['lng'])
    return distance.geodesic(city_coordinates,city_needed_coordinates).km

result = country_df.apply(calculate_distance_from,axis=1)
result
```

```python
country_df['distance'] = result
country_df
```

```python
filtered = country_df[['city_ascii','distance']]
filtered
```

```python
filtered = filtered.rename(columns = {'city_ascii': 'city'})
filtered
```

```python
output_filename = 'distance_from_lucknow.csv'
output_dir = 'output'
output_path = os.path.join(output_dir, output_filename)
filtered.to_csv(output_path, index=False)
print('Successfully written output file at {}'.format(output_path))
```

```python
other_cities = country_df[country_df['distance'] > 0]
sorted_cities = other_cities.sort_values('distance')

print("5 nearest cities")
print(sorted_cities[['city_ascii', 'distance']].head(5))
```

```python
other_cities = country_df[country_df['distance'] > 0]
sorted_cities = other_cities.sort_values('distance')

print("5 farthest cities")
print(sorted_cities[['city_ascii', 'distance']].tail(5))
```

```python
limit = [200, 500, 1000, 1500]
table_rows = []

for limit in limit :
    nearby = other_cities[other_cities['distance'] < limit]
    city_count = len(nearby)
    people = nearby['population'].sum()
    print(f"within {limit} km: {city_count} cities, population {people:,.0f}")
    table_rows.append({'limit_km': limit, 'cities': city_count, 'population': people})

table = pd.DataFrame(table_rows)
table.to_csv('output/distance_summary.csv', index=False)
table
```
